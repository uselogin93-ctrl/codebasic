**IF ELSE STATEMENTS**

if-else ka use decision making ke liye hota hai.

Agar condition true hoti hai toh ‘if’ block execute hota hai,
aur agar condition false hoti hai toh ‘else’ block execute hota hai.

Matlab program situation ke hisaab se decision leta hai.

Nested if ka matlab hota hai ek if ke andar dusra if.

**ELIF STATEMENT**

Elif ka full form hota hai else if

Jab hume multiple conditions check karni hoti hain, tab hum elif use karte hain.

Flow aisa hota hai:

Pehle if check hota hai
Agar wo false ho gaya, toh elif check hota hai
Agar wo bhi false ho gaya, toh next elif check hota hai
Finally else execute hota hai (agar sab false ho jaaye)

**elif ka use tab karte hain jab hume multiple options me se ek choose karna ho**

```python
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 70:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

### **LOOP MANIPULATION STATEMENTS IN PYTHON**

Loop manipulation statements are used to control the flow of loops during execution.

Loop manipulation statements modify the normal execution of loops.

The **break** statement terminates the loop completely and transfers control outside the loop.

The **continue** statement skips the current iteration and proceeds with the next iteration of the loop.

The **pass** statement acts as a null operation, allowing the program to run without executing any action.

These statements provide better control and flexibility while working with loops in Python.

**1. BREAK**

break ka use loop ko turant stop karne ke liye hota hai

Jab bhi break execute hota hai, loop wahi pe terminate ho jata hai aur control loop ke bahar chala jata hai

Matlab agar condition satisfy ho gayi, toh hum loop ko beech me hi rok sakte hain

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(1, 20):
    if i % 7 == 0:
        break
    print(i)
```

**2. CONTINUE**

continue current iteration ko skip karta hai

Matlab jis condition pe continue aayega, us iteration ka baaki code skip ho jayega aur loop next iteration pe chala jayega”

“Loop rukta nahi hai — sirf ek step skip hota hai”

```python
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
```

```python
for i in range(1, 21):
    if i % 2 == 0:
        continue
    print(i)
```

### **DIFFERENCE BETWEEN LIST AND ARRAY**

| Feature                 | List (Python)                     | Array (NumPy)                   |
| ----------------------- | --------------------------------- | ------------------------------- |
| Definition              | Built-in data structure in Python | Provided by NumPy               |
| Performance             | Slower for large data             | Faster and optimized            |
| Operations              | Need loops for calculations       | Direct operations without loops |
| Syntax                  | Simple (`[ ]`)                    | Requires NumPy (`np.array()`)   |
| Mathematical Operations | Not directly supported            | Supported (vectorized)          |
| Use Case                | General purpose programming       | Data Science, numerical work    |
| Flexibility             | More flexible                     | Less flexible but efficient     |
| Speed                   | Slower                            | Much faster                     |

**<U21**

👉 This one looks weird, but it’s simple:

U = Unicode string (text)

21 = maximum length of 21 characters

< = byte order (little-endian, usually not important for beginners)

✔️ So <U21 means:

👉 A string type that can store up to 21 characters
