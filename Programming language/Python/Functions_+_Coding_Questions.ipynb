### **Difference between Function,Module,Library**

| Feature           | Function                                      | Module                               | Library                        |
| ----------------- | --------------------------------------------- | ------------------------------------ | ------------------------------ |
| Definition        | A block of code that performs a specific task | A file containing multiple functions | A collection of modules        |
| Size              | Smallest unit                                 | Medium                               | Largest unit                   |
| Purpose           | Perform a single task                         | Group related functions              | Provide complete functionality |
| Dependency        | Independent                                   | Contains functions                   | Contains modules               |
| Real-life Analogy | A tool (hammer)                               | A toolbox                            | A full workshop                |

```python
num = 4
is_Prime = True
for i in range(2, num):
    if num % i == 0:
        is_Prime = False
        break
if is_Prime:
    print("Prime")
else:
    print("Not Prime")
```

```python
def prime_num(num):
  is_prime = True
  for i in range(2, num):
    if num % i == 0:
      is_prime = False
      break
  if is_prime:
      print("Prime")
  else:
      print("Not Prime")

prime_num(10)
```
