***Exception Handling***

```python
try:
  a=int(input("enter your no"))
  b=int(input("enter your no"))
  c=a/b
  print(c)
except ValueError:
  print("You cannot enter char")
except ZeroDivisionError:
  print("You cannot div a number by 0")
finally:
  print("End of Exception Handling !")
```

```python
class calcu():
  def __init__(self,n1,n2) :
    self.n1=int(n1)
    self.n2=int(n2)
  def div (self):
    try:
      c=self.n1/self.n2
      print(c)
    except ValueError:
      print("You cannot enter char")
    except ZeroDivisionError:
      print("You cannot div a number by 0")
    finally:
      print("End of Exception Handling !")
obj=calcu(9,0)
obj.div
```
