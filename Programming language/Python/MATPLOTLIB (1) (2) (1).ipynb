### **DOUGHNUT CHART**

💡 Concept

Modified pie chart with a hole in center

✅ Features

Cleaner look than pie chart

Can display labels in center

```python
import matplotlib.pyplot as plt   # Import Matplotlib for plotting

values = [40, 30, 20, 10]        # Data values for the chart

plt.pie(values, autopct='%1.1f%%')  # shows percentage on chart
                                    # Create a pie chart using the values
                                    # Each value represents a portion of the whole

# Add a white circle at the center to convert pie chart into doughnut chart
plt.gca().add_artist(
    plt.Circle((0, 0), 0.5, color='white')
)
#   Step-by-step:
#   plt.gca() → get current pie chart
#   .add_artist() → add a shape to it
#   plt.Circle(...) → create a circle

# (0,0) → center of the pie
# 0.5 → radius of the inner circle (controls size of hole)

plt.title("Doughnut Chart")      # Add title to the chart

plt.show()                       # Display the chart
```

```python
import matplotlib.pyplot as plt

# Outer ring
outer = [40, 35, 25]
# Inner ring
inner = [20, 10, 10, 15, 10, 35]

fig, ax = plt.subplots()

# Outer ring (with percentages)
ax.pie(outer, radius=1.0, autopct='%1.1f%%')

# Inner ring (with percentages)
ax.pie(inner, radius=0.7, autopct='%1.1f%%')

# Center hole
centre_circle = plt.Circle((0, 0), 0.4, color='white')
ax.add_artist(centre_circle)

ax.set_title("Donut Chart (Multi-layer)")

plt.show()
```

### **MAPS/CHARTS PLOTTED ON OUR DATASET**

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Dataset - Dataset.csv")
df
```

**BAR CHART**

```python
# Get top 10 students based on Marks
top10 = df.nlargest(10, "Marks")

plt.figure()

plt.barh(top10["Name"], top10["Marks"])

# Title
plt.title("Top 10 Students by Marks")

# X-axis label
plt.xlabel("Student Name")

# Y-axis label
plt.ylabel("Marks")

# Rotate x labels for better readability (VERY IMPORTANT when many names)
plt.xticks(rotation=45)

plt.show()
```

**COLOROFUL BAR CHART**

```python
import matplotlib.pyplot as plt
import random

top10 = df.nlargest(10, "Marks")

# Generate random colors
colors = []
for i in range(len(top10)):
    colors.append((random.random(), random.random(), random.random()))

plt.figure()

plt.bar(top10["Name"], top10["Marks"], color=colors)

plt.title("Top 10 Students by Marks")
plt.xlabel("Student Name")
plt.ylabel("Marks")

plt.xticks(rotation=45)

plt.show()
```

### **HISTOGRAM**

```python
plt.hist(df["Marks"], bins=range(0, 101, 10), edgecolor='black')

plt.title("Marks Distribution")
plt.xlabel("Marks Range")
plt.ylabel("Frequency")

plt.show()
```

### **SCATTER PLOT**

```python
filtered = df[df["Marks"] > 50]

plt.scatter(filtered["Study_Hours"], filtered["Marks"])

plt.title("Study Hours vs Marks (Above 50)")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

**CONCLUSION**

From this graph, we do not see a strong relationship between study hours and marks

There is a very weak or no correlation between study hours and marks

Study hours alone cannot explain student performance — other factors also matter

### **BOX PLOT**

```python
# Ensure numeric
df["Marks"] = pd.to_numeric(df["Marks"], errors='coerce')

# Remove nulls
clean_marks = df["Marks"].dropna()

plt.boxplot(clean_marks)

plt.title("Marks Distribution (Box Plot)")
plt.ylabel("Marks")

plt.show()
```

### **PIE CHART**

```python
import matplotlib.pyplot as plt   # Import Matplotlib for plotting

# Count how many times each gender appears in the dataset
gender_count = df["Gender"].value_counts()
# Example output:
# Male      60
# Female    40

# Create pie chart
plt.pie(
    gender_count,                 # values (counts of each gender)
    labels=gender_count.index,    # labels (Male, Female)
    autopct='%1.1f%%'             # show percentage with 1 decimal place
)

# Add title to chart
plt.title("Gender Distribution")

# Display the chart
plt.show()
```

**Pie Chart (with explode)**

```python
import matplotlib.pyplot as plt

# Count occurrences of each city
city_count = df["City"].value_counts()

# Explode first slice (optional)
explode = [0.1] + [0]*(len(city_count)-1)

plt.pie(
    city_count,                 # values
    labels=city_count.index,    # city names
    autopct='%1.1f%%',          # show %
    explode=explode,
    shadow=True
)

plt.title("City Distribution")

plt.show()
```

**Customized Pie Chart (Colors + Legend + Explode)**

```python
import matplotlib.pyplot as plt

# Count occurrences of each subject
subject_count = df["Subject"].value_counts()

# Labels (subject names)
labels = subject_count.index

# Values (counts)
sizes = subject_count.values

# Colors (can be adjusted based on number of subjects)
colors = ['orange', 'cyan', 'brown', 'gray', 'purple', 'lightgreen']

# Explode (highlight first few slices dynamically)
explode = [0.1 if i < 3 else 0 for i in range(len(labels))]

plt.pie(
    sizes,
    labels=labels,
    colors=colors[:len(labels)],   # match number of subjects
    explode=explode,
    autopct='%1.1f%%',
    shadow=True
)

plt.legend(title="Subjects", loc="center right")

plt.title("Subject Distribution")

plt.show()
```
