```python
class Game():
  def __init__(self,name,game_name, ref=0):
    self.Name=name
    self.gn=game_name
    self.__score=ref
  def kick(self):
    self.__score+=10
  def punch(self):
    self.__score+=5
  def slap(self):
    self.__score+=2
  def defan(self):
    self.__score-=5
  def shoot(self):
    self.__score-=self.__score
  def display_score(self):
    return f"score is {self.__score}"
obj=Game("Raj","Roy",5)
obj.kick()
obj.punch()
obj.slap()
obj.defan()
obj.shoot()
obj.display_score()
```
