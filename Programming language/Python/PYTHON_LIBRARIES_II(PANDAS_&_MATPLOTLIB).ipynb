### **DATA CLEANING**

**1)Remove Null Values**

**For Checking Null Values:**



```python
df.isnull().sum()
```

**A]For all columns**

| Name | Marks | Age |
| ---- | ----- | --- |
| A    | 80    | 20  |
| B    | NaN   | 22  |
| C    | 70    | NaN |


```python
df.dropna()
```

**RESULT**

| Name | Marks | Age |
| ---- | ----- | --- |
| A    | 80    | 20  |


**B]DROP IF ALL VALUES ARE NULL**



```python
df.dropna(how='all')
```

**C]Drop based on specific column**

```python
df.dropna(subset=["Marks"])
```

**D]Drop columns instead of rows**

```python
df.dropna(axis=1)
```

**E]DROPPING column only if ALL values are null**

```python
df.dropna(axis=1, how='all')
```

**F]DROP SELECTED COLUMN**

```python
df.drop("Target", axis=1)
```

**2)Fill Missing Values**

**In this Stage The Blank Values will be filled by 0**

```python
df.fillna(0)
```

**Fill specific column**

```python
df["Marks"].fillna(50)
```

**Fill with mean**

```python
df["Marks"].fillna(df["Marks"].mean())
```

### **3)Removing Duplicates**

```python
df.drop_duplicates()
```

**Remove duplicates based on column**

### **MERGING DATA**

```python
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["A", "B", "C"]
})

df2 = pd.DataFrame({
    "id": [2, 3, 4],
    "marks": [80, 90, 100]
})

pd.merge(df1, df2, on="id")
```
